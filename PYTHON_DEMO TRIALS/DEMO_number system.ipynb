## -: Decimal to Hexa-decimal :-

```python
hex1=hex(dec1)
print(dec1,"in hexi-decimal number is",hex1)
print(type(hex1))
```

```python
hex2=hex(dec2)
print(dec2,"in hexi-decimal number is",hex2)
print(type(hex2))
# float decimal can't be inputed in hex in this way.
# more advanced methods are required.
```

```python
hex3=hex(dec3)
print(dec3,"in hexi-decimal number is",hex3)
print(type(hex3))
```

## -: Hexa-decimal to Decimal :-

```python
print(int(hex3,16))     # with prefix; with base value
print(int("a",16))      # wihout prefix; with base value
print(int(hex3,0))      # with prefix; 0 base value
```

```python
print(int(hex1,16))     # with prefix; with base value
print(int("13A",16))      # wihout prefix; with base value
print(int(hex1,0))      # with prefix; 0 base value
```
