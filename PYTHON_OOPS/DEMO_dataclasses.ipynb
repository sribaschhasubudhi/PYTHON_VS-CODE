```python
from dataclasses import dataclass
```

```python
@dataclass
class Country:
    name:str
    states:int
    ut:int
    def all_states(self):
        print(f"{self.name} has {self.states} states")
my_country=Country("India",28,8)
print(my_country.name)
my_country.all_states()
```

```python
@dataclass
class Bottle:
    material:str
    quantity:float
    shape:str="Cylindrical"     # Providing a default value for shape of the bottle
    def custom_bottle(self):
        print(f"I have a {self.material} bottle which is of {self.shape} shape")
bisleri=Bottle("PLastic",1)
milton=Bottle("Steel",1.5,"Cuboidal")

print(bisleri.shape)
print(milton.quantity)
milton.custom_bottle()
milton.material="Cheap plastic"     # This is wrong. Data can be manipulated easily
print(f"Milton uses {milton.material} to make bottles. Don't buy")
```

```python
@dataclass(frozen=True)
class Pencilbox:
    pencil:int
    pen:int
    eraser:int=1
    sharpener:int=1
aman=Pencilbox(2,2,3)           # This values are permanently fixed
print("Aman's pencil box has",aman.eraser,"! He stole them!")
'''aman.pencil=0'''                   # FrozenInstanceError; You can not change the value now
```

```python
#Inheritance Example:-
@dataclass
class Department:
    division:str

@dataclass
class Cell(Department):
    cellno:int
    workers:int
    def identification(self):
        print(f"There are {self.workers} workers in this cell.")
        print(f"Ram is in {self.cellno}th cell of {self.division}-division.")

a=Department("A")
ram=Cell("A",4,24)
print(ram.cellno)
ram.identification()
```

```python
# Nested Dataclasses Example:-
@dataclass
class Address:
    street:str
    city:str
    zip_code:int

@dataclass
class Person:
    name:str
    age:int
    address:Address
loc=Address("Jane Street","New York",12345)
you=Person("Elise","23",loc)
print(you.name)
print(you.address)
print(loc.street)
```
