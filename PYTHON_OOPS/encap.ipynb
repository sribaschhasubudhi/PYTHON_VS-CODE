```python
class Student:
    def __init__(self):
        self.__name = "Alice"

    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self, new_name):

        if len(new_name) == 0:
            print("Invalid name")
            return

        self.__name = new_name
s=Student()
print(s.name)
s.name = ""
print(s.name)
```
