```python
def check(s1,s2=8,s3=12):
    r1=s3>s2
    r2=s1>s2 and s2<s3
    r3=not(s1==s2 or s2!=s3)
    # You can not write "r3=!" and "s2=!s3" because it is not valid    
    print("r1=",r1,"r2=",r2,"r3=",r3)
```

```python
check(34,34,7)
# r1:- 7>34= False
# r2:- 34>34 and 34<7= False and False= 0*0= False
# r3:- not(True or True)= not(True)= False
```
